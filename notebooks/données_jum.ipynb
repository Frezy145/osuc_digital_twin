```python
import pandas as pd 
```

```python
#  Chargement des fichiers
files = [
    "C:/Users/mirga/OneDrive/Documents/jum_num/data_2025_10_06.csv",
    "C:/Users/mirga/OneDrive/Documents/jum_num/data_2025_10_13.csv",
    "C:/Users/mirga/OneDrive/Documents/jum_num/data_2025_10_20.csv"
]

dfs = []
for f in files:
    df = pd.read_csv(f, sep=';')
```

```python
 # Conversion de la colonne date
df['date'] = pd.to_datetime(df['date'])
 # Suppression des colonnes complètement vides
df = df.dropna(axis=1, how='all')
```

```python
 # Interpoler les valeurs manquantes (méthode temporelle) 
df = df.sort_values('date').set_index('date')
df = df.interpolate(method='time')
dfs.append(df.reset_index())
```

```python
df = df.interpolate(method='time')
df = df.fillna(method='bfill').fillna(method='ffill')
```

```python
# Concaténer les trois datasets 
df_final = pd.concat(dfs, ignore_index=True)
df_final = df_final.sort_values('date')
```

```python
df_final['date'] = pd.to_datetime(df_final['date'], errors='coerce')
df_final = df_final.set_index('date')
df_final = df_final.interpolate(method='time')
df_final = df_final.fillna(method='ffill').fillna(method='bfill')
df_final = df_final.reset_index()
```

```python
na_summary = df_final.isna().sum()
print("📊 Nombre de valeurs manquantes par colonne :")
print(na_summary[na_summary > 0])
```

```python
# Sauvegarder le résultat final
df_final.to_csv("C:/Users/mirga/OneDrive/Documents/jum_num/data_combi_interpolee.csv", index=False)

print("✅ Données interpolées et combinées sauvegardées dans 'data_combinee_interpolee.csv'")
print("Nombre de lignes :", len(df_final))
print("Nombre de colonnes :", df_final.shape[1])
```
